```python
# Define the path to the training folder
training_dir = 'ASL_Dataset/Train'

# Function to display images from each subfolder
def display_dataset(folder_path, num_images=3):
    # Iterate through each subfolder (a-z) in the training directory
    for subfolder in sorted(os.listdir(folder_path)):
        subfolder_path = os.path.join(folder_path, subfolder)
        
        # Check if the subfolder is a directory
        if os.path.isdir(subfolder_path):
            # Get all file names within the subfolder
            files = os.listdir(subfolder_path)
            
            # Limit to the first few files
            files = files[:num_images]
            
            # Display a few images from this subfolder
            print(f"Displaying images from folder: {subfolder}")
            
            # Create a plot with num_images columns
            plt.figure(figsize=(15, 5))
            
            for i, file in enumerate(files):
                file_path = os.path.join(subfolder_path, file)
                
                # Open image
                img = Image.open(file_path)
                
                # Display image in a subplot
                plt.subplot(1, num_images, i + 1)
                plt.imshow(img)
                plt.title(f'{subfolder} - Image {i + 1}')
                plt.axis('off')
        
        # This makes it able to show the images. 
        plt.show()

# Call the function to display images
display_dataset(training_dir, num_images=3)

```

Establishing the location to the folder holding your training photos is the first step in preparing images for dimensionality reduction using PCA and t-SNE. After loading each image, decrease its complexity by resizing it to a standard size and converting it to greyscale. Then, in order to use each processed image as input for additional analysis, flatten it into a one-dimensional vector. Compile these flattened vectors into a dataset, then transform it into numpy arrays so that it may be used with the t-SNE and PCA methods for effective dimensionality reduction and visualization.

```python
# Define the path to the training folder
training_dir = 'ASL_Dataset/Train'

# Function to load images and convert them into flat vectors for PCA and t-SNE
def load_images_for_pca(folder_path, num_images_per_class=100, image_size=(64, 64)):
    image_data = []
    labels = []
    
    for subfolder in sorted(os.listdir(folder_path)):
        subfolder_path = os.path.join(folder_path, subfolder)
        
        if os.path.isdir(subfolder_path):
            files = os.listdir(subfolder_path)[:num_images_per_class]
            
            for file in files:
                file_path = os.path.join(subfolder_path, file)
                
                # Load, resize, and convert image to grayscale
                img = Image.open(file_path).convert('L').resize(image_size)
                
                # Flatten image to a vector and add to dataset
                img_array = np.array(img).flatten()
                image_data.append(img_array)
                labels.append(subfolder)  # Use the subfolder name as the label

    # Convert lists to numpy arrays
    return np.array(image_data), np.array(labels)

image_data, labels = load_images_for_pca(training_dir, num_images_per_class=100, image_size=(64, 64))
```

To ensure reproducibility, start by setting a random seed. Define any necessary parameters, then create a random sample from the filtered dataset for analysis. Next, apply PCA to the sampled data to reduce its dimensionality and print the explained variance to evaluate how much information is retained after this step. Following PCA, apply t-SNE to further reduce the data and make it suitable for visualization. Encode any categorical labels as numerical values for consistency. Finally, plot the t-SNE results, using the encoded labels to distinguish between different groups in the visualization. This sequence provides a streamlined approach for visualizing high-dimensional data with both PCA and t-SNE.

```python
# Set a random seed for reproducibility
np.random.seed(42)

# Parameters
sample_size = 1000 
pca_components = 50  

filtered_indices = [i for i, label in enumerate(labels) if label not in ('J', 'space')]
filtered_images = image_data[filtered_indices]
filtered_labels = labels[filtered_indices]

# Create a random sample from the filtered data
random_indices = np.random.choice(len(filtered_images), size=sample_size, replace=False)
subset_images = filtered_images[random_indices]
subset_labels = filtered_labels[random_indices]

# Step 2: Apply PCA on the sampled data
pca = PCA(n_components=pca_components, random_state=42)
subset_images_reduced = pca.fit_transform(subset_images)

# Print explained variance to check information retention
explained_variance = np.sum(pca.explained_variance_ratio_) * 100
print(f"PCA explained variance for {pca_components} components: {explained_variance:.2f}%")

# Step 3: Apply t-SNE for visualization on the PCA-reduced data
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
subset_images_tsne = tsne.fit_transform(subset_images_reduced)

# Encode labels to numerical values
label_encoder = LabelEncoder()
numeric_labels = label_encoder.fit_transform(subset_labels)

# Step 4: Plot the t-SNE result with labels
plt.figure(figsize=(10, 10))
scatter = plt.scatter(subset_images_tsne[:, 0], subset_images_tsne[:, 1], c=numeric_labels, cmap="tab20", s=5, alpha=0.7)
plt.colorbar(scatter, label='Class Labels')
plt.title("t-SNE Visualization of ASL Dataset after PCA (Random Sample)")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.show()
```

```python
def create_dataset(dataset_path, img_size=(28, 28)):
    X = []
    y = []
    labels = sorted(os.listdir(dataset_path))
    label_map = {label: idx for idx, label in enumerate(labels)}

    for label in labels:
        label_path = os.path.join(dataset_path, label)
        if os.path.isdir(label_path):
            for file in os.listdir(label_path):
                if not file.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff')):
                    print(f"Skipping non-image file: {file}")
                    continue
                file_path = os.path.join(label_path, file)
                try:
                    img = preprocess_image_for_cnn(file_path, img_size)
                    X.append(img)
                    y.append(label_map[label])
                except ValueError as e:
                    print(e)
                    continue

    X = np.array(X).reshape(-1, img_size[0], img_size[1], 1)  # Add channel dimension
    y = np.array(y)
    return X, y, label_map
```

my build_cnn_model method constructs a convolutional neural network (CNN) for image classification.

```python
def build_cnn_model(input_shape, num_classes):
    model = Sequential([
        # Convolutional layers
        Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        MaxPooling2D((2, 2)),

        Conv2D(64, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),

        Conv2D(128, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),

        # Flatten the feature map
        Flatten(),

        # Fully connected layers
        Dense(128, activation='relu'),
        Dropout(0.5),
        Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model
```

```python
def evaluate_model_per_class(model, X_test, y_test, label_map, images_per_class=50):

    # Reverse the label map to get class names
    reverse_label_map = {idx: label for label, idx in label_map.items()}
    
    # Group images by their true label
    class_indices = {label: np.where(y_test == idx)[0] for label, idx in label_map.items()}
    
    overall_accuracy = []
    
    for label, indices in class_indices.items():
        # Select a fixed number of images for the class
        if len(indices) < images_per_class:
            print(f"Not enough images for class '{label}'. Using {len(indices)} images.")
            selected_indices = indices
        else:
            selected_indices = np.random.choice(indices, size=images_per_class, replace=False)
        
        # Get the selected images and labels
        selected_images = X_test[selected_indices]
        true_labels = y_test[selected_indices]
        
        # Predict class probabilities
        predictions = model.predict(selected_images)
        predicted_labels = np.argmax(predictions, axis=1)
        
        # Calculate accuracy for this class
        class_accuracy = np.mean(predicted_labels == true_labels)
        overall_accuracy.append(class_accuracy)
        
        # Print results for this class
        print(f"Class '{label}' - Accuracy: {class_accuracy * 100:.2f}%")
        
        # Visualize a few images from this class
        plt.figure(figsize=(10, 5))
        for i in range(min(10, len(selected_images))):  # Show up to 10 images per class
            plt.subplot(2, 5, i + 1)
            plt.imshow(selected_images[i].reshape(28, 28), cmap='gray')  # Adjust size to match input
            true_label = reverse_label_map[true_labels[i]]
            predicted_label = reverse_label_map[predicted_labels[i]]
            color = 'green' if true_label == predicted_label else 'red'
            plt.title(f"True: {true_label}\nPred: {predicted_label}", color=color)
            plt.axis('off')
        plt.tight_layout()
        plt.show()
    
    # Compute overall accuracy
    mean_accuracy = np.mean(overall_accuracy)
    print(f"Overall accuracy across all classes: {mean_accuracy * 100:.2f}%")
```

```python
def analyze_model_predictions(model, X_test, y_test, label_map, img_size=(28, 28), num_predictions=15, num_per_class=8):
    
    # Convert labels if one-hot encoded
    if len(y_test.shape) > 1:
        y_test = np.argmax(y_test, axis=1)
    
    # Reverse label map for display
    reverse_label_map = {v: k for k, v in label_map.items()}
    
    # Overall predictions visualization
    print("Overall Prediction Analysis:")
    predictions = model.predict(X_test[:num_predictions])
    plt.figure(figsize=(15, 10))
    for i in range(num_predictions):
        plt.subplot(3, 5, i + 1)
        plt.imshow(X_test[i].reshape(img_size), cmap='gray')
        true_label = y_test[i]
        pred_label = np.argmax(predictions[i])
        is_correct = true_label == pred_label
        confidence = predictions[i][pred_label] * 100
        color = 'green' if is_correct else 'red'
        plt.title(f"True: {reverse_label_map[true_label]}\nPred: {reverse_label_map[pred_label]}\nConf: {confidence:.2f}%",
                  color=color, fontsize=8)
        plt.axis('off')
    plt.tight_layout()
    plt.show()
    
    # Per-class predictions visualization
    print("Per-Class Prediction Analysis:")
    class_samples = {cls: 0 for cls in range(len(label_map))}
    plt.figure(figsize=(15, len(label_map) * 2))
    for i, (img, true_label) in enumerate(zip(X_test, y_test)):
        if class_samples[true_label] < num_per_class:
            prediction = model.predict(img.reshape(1, *img_size, 1))
            pred_label = np.argmax(prediction)
            confidence = prediction[0][pred_label] * 100
            is_correct = true_label == pred_label
            color = 'green' if is_correct else 'red'
            
            plt.subplot(len(label_map), num_per_class, true_label * num_per_class + class_samples[true_label] + 1)
            plt.imshow(img.reshape(img_size), cmap='gray')
            plt.title(f"True: {reverse_label_map[true_label]}\nPred: {reverse_label_map[pred_label]}\nConf: {confidence:.2f}%",
                      color=color, fontsize=8)
            plt.axis('off')
            class_samples[true_label] += 1
        if all(count == num_per_class for count in class_samples.values()):
            break
    plt.tight_layout()
    plt.show()
    
    # Evaluate accuracy per class
    print("Class-Wise Accuracy Analysis:")
    evaluate_model_per_class(model, X_test, y_test, label_map, images_per_class=num_per_class)


analyze_model_predictions(model, 
                          X_test, 
                          y_test, 
                          label_map, 
                          img_size=(40, 40), 
                          num_predictions=15, 
                          num_per_class=8)
```
